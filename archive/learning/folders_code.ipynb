```python
#Folders

import os

# Ruta donde están los archivos
base_path = '/data/projects/Chemdocking/dock_score/destabilizer_with_poses/pose_1'

final= '/projectbig/jupyternotebook/jrivera/output/pocket_results/ligand_poses/destabilizers'

# Itera sobre cada archivo en la carpeta
for filename in os.listdir(base_path):
    if "_" in filename:
        parts = filename.split('_')
        if len(parts) > 4:
            # Obtiene la 5ta parte del nombre
            folder_name = parts[4]
            # Ruta del nuevo directorio
            new_dir = os.path.join(final, folder_name)
            # Crea el directorio si no existe
            if not os.path.exists(new_dir):
                os.mkdir(new_dir)
```
