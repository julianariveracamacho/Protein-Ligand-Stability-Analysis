# 07 · Compound-level interaction statistics

Fixed presence rule, compound-level Fisher tests, effect sizes and a single Benjamini–Yekutieli multiple-testing family. The default input is synthetic. This replaces the earlier minimum-p-value threshold scan; it does not retroactively validate historical results.

## Select input explicitly

```python
from pathlib import Path
import sys
root = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'analysis/interactions.py').is_file()), None)
if root is None: raise FileNotFoundError('Start within repository')
sys.path.insert(0, str(root))
from analysis.interactions import load_counts, summarize, compare_groups
input_path = root / 'examples/synthetic/output/synthetic_pose_counts.csv'
# For authorized research data, explicitly set input_path to the notebook 06 pose_counts.csv.
# Assign independent compound groups and standardize pose coverage before testing.
data = load_counts(input_path)
means, presence = summarize(data)
print('Statistical units:', len(presence), 'compounds')
```

## Run all pre-defined comparisons

```python
results = compare_groups(presence, control='control', groups=('stabilizer', 'destabilizer'))
results
```

## Save the complete test family

```python
output_dir = root / 'results'
output_dir.mkdir(exist_ok=True)
results.to_csv(output_dir / 'compound_statistics.csv', index=False)
# Interpret effect sizes and adjusted p-values together. No significance-based filtering.
# Synthetic values are demonstrations only; research data need an agreed analysis plan.
```
